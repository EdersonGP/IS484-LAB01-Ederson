## 5. Ejercicio práctico de laboratorio

### 5.1 Planteamiento del problema

Se desea automatizar el control de temperatura de un ambiente mediante un sensor y un sistema de climatización. El agente debe decidir qué acción realizar según la temperatura actual.

### 5.2 Diseño del agente - Ficha PEAS
| Elemento | Descripción |
|---|---|
| **Percepción (S)** | Temperatura actual medida por el sensor. |
| **Acciones (A)** | Encender aire acondicionado, encender calefacción o mantener el sistema sin cambios. |
| **Entorno (E)** | Ambiente o habitación cuya temperatura se desea controlar. |
| **Objetivo** | Mantener la temperatura cercana a 22 °C. |
| **Medida de desempeño (P)** | Mantener la temperatura dentro del margen establecido, evitando cambios innecesarios y procurando un consumo eficiente de energía. |

### 5.3 Reglas de decisión
| Condición | Acción a tomar |
|---|---|
| Si la temperatura está muy por encima de la deseada | Encender aire acondicionado |
| Si la temperatura está muy por debajo de la deseada | Encender calefacción |
| Si la temperatura está cerca de la deseada | Mantener la climatización sin cambios |

```python
def agente_termostato(
    temperatura_actual,
    temperatura_objetivo=22,
    margen=1.0
):
    """
    Agente reactivo simple.

    Percepción:
    Temperatura actual del ambiente.

    Acciones:
    Encender aire acondicionado,
    encender calefacción
    o mantener la climatización.

    Entorno:
    Ambiente cuya temperatura se desea controlar.

    Objetivo:
    Mantener la temperatura cercana a 22 °C.

    Medida de desempeño:
    Mantener la temperatura dentro del margen definido
    con la menor cantidad de acciones innecesarias.
    """

    diferencia = temperatura_actual - temperatura_objetivo

    if diferencia > margen:
        return "Encender aire acondicionado"

    elif diferencia < -margen:
        return "Encender calefaccion"

    else:
        return "Mantener temperatura"
```

```python
### 5.5 Simulación del agente
percepciones = [
    18.0,
    19.5,
    21.8,
    24.3,
    26.0,
    23.0,
    20.5,
    22.5,
    21.0
]

for t in percepciones:
    accion = agente_termostato(t)

    print(
        f"Percepcion: {t} C "
        f"-> Accion del agente: {accion}"
    )

```

### 5.6 Reflexión

**¿Este agente termostato es racional?**

Sí. El agente es racional porque, a partir de la temperatura percibida, selecciona una acción que busca mantener el ambiente cercano a la temperatura objetivo de 22 °C.

**¿Bajo qué condiciones dejaría de ser racional?**

Podría dejar de ser racional si el sensor entrega mediciones incorrectas, si el sistema de calefacción o aire acondicionado falla, o si el margen definido es demasiado amplio o inadecuado. En estas situaciones, el agente podría tomar una acción que no permita mantener correctamente la temperatura deseada.

## 5.9 Extensión para la casa

### Agente de riego automático

El sistema debe decidir si regar o no un jardín considerando dos percepciones:

- `humedad_suelo`: valor entre 0 y 100.
- `llovera_hoy`: `True` o `False` según el pronóstico del tiempo.

### Parte A: Ficha PEAS
| Elemento | Descripción |
|---|---|
| **Percepción (S)** | Humedad actual del suelo y pronóstico de lluvia para el día. |
| **Acciones (A)** | Regar o no regar el jardín. |
| **Entorno (E)** | Jardín, suelo, plantas y condiciones climáticas. |
| **Objetivo** | Mantener una humedad adecuada en el suelo evitando riego innecesario. |
| **Medida de desempeño (P)** | Mantener el suelo con humedad suficiente utilizando la menor cantidad de agua posible y evitando regar cuando se espera lluvia. |

### Parte B: Reglas de decisión

Para este agente se considera que el suelo está seco cuando la humedad es menor al 30 %.
| Condición | Acción |
|---|---|
| Humedad menor a 30 % **y** no lloverá hoy | Regar |
| Humedad menor a 30 % **y** lloverá hoy | No regar |
| Humedad igual o mayor a 30 % | No regar |

### Parte B: Reglas de decisión

Para este agente se considera que el suelo está seco cuando la humedad es menor al 30 %.
| Condición | Acción |
|---|---|
| Humedad menor a 30 % **y** no lloverá hoy | Regar |
| Humedad menor a 30 % **y** lloverá hoy | No regar |
| Humedad igual o mayor a 30 % | No regar |

### Parte C: Implementación del agente

```python
def agente_riego(humedad_suelo, llovera_hoy):
    """
    Agente reactivo simple para riego automático.

    Percepciones:
    - Humedad del suelo.
    - Pronóstico de lluvia.

    Acción:
    - Regar.
    - No regar.

    Objetivo:
    Mantener una humedad adecuada evitando desperdiciar agua.
    """

    if humedad_suelo < 30 and not llovera_hoy:
        return "Regar"

    else:
        return "No regar"
```

### Parte D: Pruebas del agente

```python
pruebas = [
    (20, False),   # Suelo seco y no lloverá
    (20, True),    # Suelo seco pero lloverá - caso de conflicto
    (45, False),   # Humedad adecuada y no lloverá
    (60, True),    # Suelo húmedo y lloverá
    (29, False),   # Cerca del límite, pero todavía seco
    (30, False)    # Caso límite exacto
]

for humedad, lluvia in pruebas:

    accion = agente_riego(
        humedad,
        lluvia
    )

    print(
        f"Humedad: {humedad}% | "
        f"Llovera hoy: {lluvia} "
        f"-> Accion: {accion}"
    )
```

### Parte E: Reflexión

**¿Qué tan distinto fue diseñar este agente respecto al termostato?**

El agente de riego requiere analizar dos percepciones al mismo tiempo: la humedad del suelo y el pronóstico de lluvia. En cambio, el termostato solo utiliza una percepción, que es la temperatura actual. Por ello, las reglas del agente de riego necesitan combinar condiciones antes de tomar una decisión.

**¿A qué tipo de agente corresponde?**

Corresponde a un agente reactivo simple porque toma una decisión únicamente a partir de las percepciones actuales y no almacena información ni recuerda estados anteriores.
